```python
import numpy as np 
import matplotlib.pyplot as plt
import json
import glob
```

```python
# get paths
metric_path_maml_ml1 = "metrics/ML1/ml1.json"
metric_path_maml_ml10 = "metrics/ML10/ml10.json"
```

```python
# get data
with open(metric_path_maml_ml1, 'r') as f:
    ml1_data = json.load(f)
with open(metric_path_maml_ml10, 'r') as f:
    ml10_data = json.load(f)
```

```python
# plot ML1
iters_ml1 = ml1_data["iterations"]
pre_reward_ml1 = ml1_data["pre_mean"]
post_reward_ml1 = ml1_data["post_mean"]
gap_ml1 = ml1_data["gap"]
plt.plot(iters_ml1, pre_reward_ml1, label="Pre reward")
plt.plot(iters_ml1, post_reward_ml1, label="Post reward")
plt.plot(iters_ml1, gap_ml1, label="gap")
plt.xlabel("Meta Iterations")
plt.ylabel("Reward")
plt.legend()
plt.title("ML1 reward vs iteration")
plt.show()
```

```python
# plot ML10
iters_ml10 = ml10_data["iterations"]
pre_reward_ml10 = ml10_data["pre_mean"]
post_reward_ml10 = ml10_data["post_mean"]
gap_ml10 = ml10_data["gap"]
plt.plot(iters_ml10, pre_reward_ml10, label="Pre reward")
plt.plot(iters_ml10, post_reward_ml10, label="Post reward")
plt.plot(iters_ml10, gap_ml10, label="gap")
plt.xlabel("Meta Iterations")
plt.ylabel("Reward")
plt.legend()
plt.title("ML10 reward vs iteration")
plt.show()
```
